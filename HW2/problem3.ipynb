```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
# import train dataset
train = pd.read_csv('train.csv')
train.head()
```

```python
# import test dataset
test = pd.read_csv('test.csv')
test.head()
```

```python
# TRAIN:
# ignore columns: id, date and zipcode
# also ignore price column because this is the response and what we are trying to predict
columns_to_drop = ['Unnamed: 0', 'zipcode', 'price']
x_train = train.drop(columns=columns_to_drop)
x_train.head()
```

```python
# TEST:
# ignore columns: id, date and zipcode
# also ignore price column because this is the response and what we are trying to predict
columns_to_drop = ['Unnamed: 0', 'id', 'date', 'zipcode', 'price']
x_test = test.drop(columns=columns_to_drop)
x_test.head()
```

```python
# scale the data so that each feature has a mean of 0 and standard deviation of 1
scaler = StandardScaler()

# scale train dataset
for col in x_train:
    train_scaled = x_train[col].values.reshape(-1, 1) #reshape
    x_train[col] = scaler.fit_transform(train_scaled) 

# scale test dataset
for col in x_test:
    test_scaled = x_test[col].values.reshape(-1, 1) #reshape
    x_test[col] = scaler.fit_transform(test_scaled) 


# check mean and standard deviation of each feature to double check that scaling worked
for feature in x_train:
    mean = x_train.mean()
    std = x_train.var()

for feature in x_test:
    mean = x_test.mean()
    std = x_test.var()

# organize all variables into a table
stats_train = {
    'mean': mean,
    'std': std
}

stats_test = {
    'mean': mean,
    'std': std
}

feature_data_train = pd.DataFrame(stats_train)
feature_data_test = pd.DataFrame(stats_test)
feature_data_test
```

```python
# divide the price by 1000 for all rows in the train dataset
y_train = train['price'] / 1000
```

```python
# divide the price by 1000 for all rows in the test dataset
y_test = test['price'] / 1000
```

```python
# need to organize training data features into a matrix:
matrix_train = x_train.values

price_matrix_train = y_train.values
```

```python
# need to organize features into a matrix for test:
matrix_test = x_test.values

price_matrix_test = y_test.values
```

```python
# include a column of ones into features for the bias
# lecture slides showed column of ones in features matrix:

# train dataset:
shape = matrix_train.shape[0]
matrix_train = np.append(matrix_train, np.ones((shape,1)), axis=1)
 
# test dataset:
shape_test = matrix_test.shape[0]
matrix_test = np.append(matrix_test, np.ones((shape,1)), axis=1)
```

```python
# theta = ((x^T)*x)^-1 * (x^T)y

def closed_form_calc(matrix, price_matrix):

    x_t = matrix.T @ matrix

    x_t_inv = np.linalg.pinv(x_t) #pinv because singular

    x_t_y = matrix.T @ price_matrix

    theta = x_t_inv @ x_t_y

    return theta

theta_train = closed_form_calc(matrix_train, price_matrix_train)
theta_train
```

```python
# Write a function to predict the response on a new testing point.

def predict_response(x, theta):
    return x @ theta

pred = predict_response(matrix_train, theta_train)
```

```python
# find MSE and r^2 value for train:

MSE_train = mean_squared_error(price_matrix_train, pred)
r2_train = r2_score(price_matrix_train, pred)

print('MSE metric: ', MSE_train)
print('R squared metric: ', r2_train)
```

```python
# calculate theta for test dataset
theta_test = closed_form_calc(matrix_test, price_matrix_test)
# find prediction 
pred_test = predict_response(matrix_test, theta_test)

# find MSE and r^2 value for test:
MSE_train = mean_squared_error(price_matrix_test, pred_test)
r2_train = r2_score(price_matrix_test, pred_test)

print('MSE metric: ', MSE_train)
print('R squared metric: ', r2_train)
```
