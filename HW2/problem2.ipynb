```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
# import house_data dataset
original_dataset = pd.read_csv('kc_house_data.csv')
original_dataset.head()
```

```python
# import train dataset
train = pd.read_csv('train.csv')
train.head()
```

```python
# import test dataset
test = pd.read_csv('test.csv')
test.head()
```

```python
# TRAIN:
# ignore columns: id, date and zipcode
# also ignore price column because this is the response and what we are trying to predict
columns_to_drop = ['Unnamed: 0', 'zipcode', 'price']
x_train = train.drop(columns=columns_to_drop)
x_train.head()
```

```python
# TEST:
# ignore columns: id, date and zipcode
# also ignore price column because this is the response and what we are trying to predict
columns_to_drop = ['Unnamed: 0', 'id', 'date', 'zipcode', 'price']
x_test = test.drop(columns=columns_to_drop)
x_test.head()
```

```python
# scale the data so that each feature has a mean of 0 and standard deviation of 1
scaler = StandardScaler()

# scale train dataset
for col in x_train:
    train_scaled = x_train[col].values.reshape(-1, 1) #reshape
    x_train[col] = scaler.fit_transform(train_scaled) 

# scale test dataset
for col in x_test:
    test_scaled = x_test[col].values.reshape(-1, 1) #reshape
    x_test[col] = scaler.fit_transform(test_scaled) 


# check mean and standard deviation of each feature to double check that scaling worked
for feature in x_train:
    mean = x_train.mean()
    std = x_train.var()

for feature in x_test:
    mean = x_test.mean()
    std = x_test.var()

# organize all variables into a table
stats_train = {
    'mean': mean,
    'std': std
}

stats_test = {
    'mean': mean,
    'std': std
}

feature_data_train = pd.DataFrame(stats_train)
feature_data_test = pd.DataFrame(stats_test)
feature_data_test
```

```python
# divide the price by 1000 for all rows in the train dataset
y_train = train['price'] / 1000
y_train
```

```python
# divide the price by 1000 for all rows in the test dataset
y_test = test['price'] / 1000
y_test
```

```python
# train multiple linear regression model on the training dataset
# using the existing python sklearn linear regression package:

model = LinearRegression()
model.fit(x_train, y_train)

# print model intercept
print("Intercept: ", model.intercept_)

# print model coefficient in a table with corresponding feature
coeff = {
    'feature': x_train.columns,
    'Coefficient': model.coef_ 
    }


pd.DataFrame(coeff)
```

```python
# find MSE and r^2 on training set

y_train_pred = model.predict(x_train)

MSE_train = mean_squared_error(y_train, y_train_pred)
r2_train = r2_score(y_train, y_train_pred)

print('MSE metric: ', MSE_train)
print('R squared metric: ', r2_train)
```

```python
# find MSE and r^2 on test set

y_test_pred = model.predict(x_test)

MSE_test = mean_squared_error(y_test, y_test_pred)
r2_test = r2_score(y_test, y_test_pred)

print('MSE metric: ', MSE_test)
print('R squared metric: ', r2_test)
```
