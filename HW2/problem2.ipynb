```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
# import house_data dataset
original_dataset = pd.read_csv('kc_house_data.csv')
original_dataset.head()
```

```python
# import training dataset
train = pd.read_csv('train.csv')
train.head()
```

```python
# ignore columns: id, date and zipcode
# also ignore price column because this is the response and what we are trying to predict
features = train.drop(columns=['Unnamed: 0', 'zipcode', 'price'])
features.head()
```

```python
# scale the data so that each feature has a mean of 0 and standard deviation of 1
scaler = StandardScaler()

for col in features:
    new_col = features[col].values.reshape(-1, 1) #reshape
    features[col] = scaler.fit_transform(new_col) 


# check mean and standard deviation of each feature to double check that scaling worked
for feature in features:
    mean = features.mean()
    std = features.var()

# organize all variables into a table
stats = {
    'mean': mean,
    'std': std
}

feature_data = pd.DataFrame(stats)
feature_data
```
