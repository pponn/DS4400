Consider a feature $X$, a response variable $Y$, and $N$ samples of training data. Implement a polynomial regression model that fits a polynomial of degree $p$ to the data using the least-square method. Use your own implementation from Problem 3 and adapt it for polynomial
regression. If $p=2$, the model will use two features ($X$ and $X^2$), if $p=3$ the model will use 3 features ($X,X^2,X^3$), and so on for larger values of $p$.

```python
#X = feature, p = degree
def polynomial_regression(X, p):

    X = X.values # to matrix
    
    X = X.reshape(-1, 1)  # columns
    features = X.copy() # this is X to the first power
    
    for i in range(2, p+1): # start from exponent 2 bc we have X above
        
        features = np.append(features, X**i, axis=1)

    shape = features.shape[0]
    features = np.append(features, np.ones((shape,1)), axis=1)  

    return features
```

```python
test = polynomial_regression(x_train['sqft_living'], 2)
#test
```

```python
# implementation from problem 3 adapted for polynomial regression:
# theta = ((x^T)*x)^-1 * (x^T)y

def closed_form_calc_polynomial(matrix, price_matrix, p):

    matrix = polynomial_regression(matrix, p)

    xT_x = matrix.T @ matrix

    xTx_inv = np.linalg.pinv(xT_x) #pinv because singular

    xT_y = matrix.T @ price_matrix

    theta = xTx_inv @ xT_y

    return theta
```

```python
x = x_test['sqft_living']
closed_form_calc_polynomial(x, price_matrix_test, 3)

x = x_train['sqft_living']
closed_form_calc_polynomial(x, price_matrix_train, 3)
```

```python
def predict_response(x, theta):
    return x @ theta
```

Consider the house price prediction problem with feature $X=$ sqft_living. Train a polynomial regression model for different values of $p \le 5$ using your implementation. Include a table with the MSE and $R^2$ metrics on both the training and testing data for at least 3 different values of $p$. Discuss your observations on how the MSE and $R^2$ metrics change with the degree of the polynomial.

```python
x = x_train['sqft_living']
ps = [0, 1, 2, 3, 4, 5]
mses = []
r_squared = []

for p in ps:
    
    poly = polynomial_regression(x, p)
    theta_poly = closed_form_calc_polynomial(x, price_matrix_train, p)

    y_pred = predict_response(poly, theta_poly)

    mses.append(mean_squared_error(price_matrix_train, y_pred))
    r_squared.append(r2_score(price_matrix_train, y_pred))


# organize all variables into a table
stats_train = {
    'p value': ps,
    'MSE metrics': mses,
    'R squared metrics': r_squared
}

print("Training data:")
data_train = pd.DataFrame(stats_train)
data_train
```

```python
x = x_test['sqft_living']
ps = [0, 1, 2, 3, 4, 5]
mses = []
r_squared = []

for p in ps:
    
    poly = polynomial_regression(x, p)
    theta_poly = closed_form_calc_polynomial(x, price_matrix_test, p)

    y_pred = predict_response(poly, theta_poly)

    mses.append(mean_squared_error(price_matrix_test, y_pred))
    r_squared.append(r2_score(price_matrix_test, y_pred))


# organize all variables into a table
stats_test = {
    'p value': ps,
    'MSE metrics': mses,
    'R squared metrics': r_squared
}

print("Test data:")
data_test = pd.DataFrame(stats_test)
data_test
```

The MSE and R squared metrics change with the degree of the polynomial, or with different p values, where the MSE values decrease with each increasnig p value and the R squared metric increases with each increasing p value. This is observed in both training and test datasets.
